```python
import pandas as pd
df = pd.read_csv("/content/netflix_titles.csv")
```

```python
print(df.shape)
print(df.isnull().sum())

for col in ['director', 'cast', 'country', 'rating', 'listed_in', 'description']:
    df[col] = df[col].fillna('')

df = df.drop_duplicates(subset='title').reset_index(drop=True)
```

```python
import matplotlib.pyplot as plt, seaborn as sns

sns.countplot(x='type', data=df); plt.title('Movies vs TV Shows'); plt.show()

df['year_added'] = pd.to_datetime(df['date_added'].str.strip(), errors='coerce').dt.year
df['year_added'].value_counts().sort_index().plot(kind='bar', figsize=(10,4))
plt.title('Titles added per year'); plt.show()

genres = df['listed_in'].str.split(', ').explode().value_counts().head(10)
genres.plot(kind='barh', title='Top 10 genres'); plt.show()
```

```python
import re

def clean_list(text, limit=None):
    items = [i.strip().replace(' ', '').lower() for i in text.split(',') if i.strip()]
    return ' '.join(items[:limit] if limit else items)

df['soup'] = (
    df['listed_in'].apply(clean_list) + ' ' +
    df['listed_in'].apply(clean_list) + ' ' +          # genres weighted x2
    df['director'].apply(clean_list) + ' ' +
    df['cast'].apply(lambda x: clean_list(x, limit=5)) + ' ' +
    df['country'].apply(clean_list, limit=None) + ' ' +
    df['type'].str.lower() + ' ' +
    df['description'].str.lower()
)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import linear_kernel

tfidf = TfidfVectorizer(stop_words='english', ngram_range=(1, 2), min_df=2)
tfidf_matrix = tfidf.fit_transform(df['soup'])
print(tfidf_matrix.shape)

cosine_sim = linear_kernel(tfidf_matrix, tfidf_matrix)
indices = pd.Series(df.index, index=df['title'].str.lower())
```

```python
import difflib

def recommend(title, n=10, same_type=False):
    key = title.lower().strip()

    # Fuzzy match if the exact title isn't found
    if key not in indices:
        close = difflib.get_close_matches(key, indices.index, n=3, cutoff=0.6)
        if not close:
            return f"'{title}' not found."
        print(f"Did you mean: {close}? Using '{close[0]}'.")
        key = close[0]

    i = indices[key]
    scores = list(enumerate(cosine_sim[i]))
    scores = sorted(scores, key=lambda x: x[1], reverse=True)[1:]

    if same_type:
        scores = [s for s in scores if df.loc[s[0], 'type'] == df.loc[i, 'type']]

    top = scores[:n]
    result = df.iloc[[s[0] for s in top]][['title', 'type', 'listed_in', 'release_year', 'rating']].copy()
    result['similarity'] = [round(s[1], 3) for s in top]
    return result

recommend("Stranger Things", n=10)
```

```python
import random

def genre_hit_rate(n_samples=200, k=10):
    hits = []
    for i in random.sample(range(len(df)), n_samples):
        query_genres = set(df.loc[i, 'listed_in'].split(', '))
        scores = sorted(enumerate(cosine_sim[i]), key=lambda x: x[1], reverse=True)[1:k+1]
        for j, _ in scores:
            rec_genres = set(df.loc[j, 'listed_in'].split(', '))
            hits.append(len(query_genres & rec_genres) > 0)
    return np.mean(hits)

import numpy as np
print(f"Genre hit rate@10: {genre_hit_rate():.2%}")
```

```python
!pip install sentence-transformers -q
from sentence_transformers import SentenceTransformer

st_model = SentenceTransformer('all-MiniLM-L6-v2')

# Text for embeddings: genres + description read naturally
texts = (df['listed_in'] + '. ' + df['description']).tolist()
emb = st_model.encode(texts, batch_size=64, show_progress_bar=True,
                      normalize_embeddings=True)

emb_sim = emb @ emb.T     # cosine similarity (vectors are normalized)
print(emb_sim.shape)
```

```python
hybrid_sim = 0.4 * cosine_sim + 0.6 * emb_sim

def recommend_with(sim_matrix, title, n=10):
    key = title.lower().strip()
    if key not in indices:
        close = difflib.get_close_matches(key, indices.index, n=1, cutoff=0.6)
        if not close:
            return f"'{title}' not found."
        key = close[0]
    i = indices[key]
    scores = sorted(enumerate(sim_matrix[i]), key=lambda x: x[1], reverse=True)[1:n+1]
    out = df.iloc[[s[0] for s in scores]][['title', 'type', 'listed_in', 'release_year']].copy()
    out['similarity'] = [round(s[1], 3) for s in scores]
    return out

recommend_with(hybrid_sim, "Stranger Things")
```

```python
# Description-only TF-IDF baseline
tfidf_desc = TfidfVectorizer(stop_words='english', ngram_range=(1, 2), min_df=2)
desc_matrix = tfidf_desc.fit_transform(df['description'])
desc_sim = linear_kernel(desc_matrix, desc_matrix)

def evaluate(sim_matrix, n_samples=300, k=10, seed=42):
    rng = random.Random(seed)
    hits, cast_hits = [], []
    for i in rng.sample(range(len(df)), n_samples):
        q_genres = set(df.loc[i, 'listed_in'].split(', '))
        top = np.argsort(sim_matrix[i])[::-1][1:k+1]
        for j in top:
            r_genres = set(df.loc[j, 'listed_in'].split(', '))
            hits.append(len(q_genres & r_genres) > 0)
    return np.mean(hits)

results = pd.DataFrame({
    'Model': ['TF-IDF (description only)', 'TF-IDF (full soup)',
              'Sentence embeddings', 'Hybrid'],
    'Genre hit rate@10': [evaluate(desc_sim), evaluate(cosine_sim),
                          evaluate(emb_sim), evaluate(hybrid_sim)]
}).round(3)
results
```

```python
from sklearn.decomposition import PCA

sample = df.sample(1500, random_state=1)
coords = PCA(n_components=2).fit_transform(emb[sample.index])
sample = sample.assign(x=coords[:, 0], y=coords[:, 1],
                       main_genre=sample['listed_in'].str.split(', ').str[0])

top_genres = sample['main_genre'].value_counts().head(8).index
plt.figure(figsize=(10, 7))
sns.scatterplot(data=sample[sample.main_genre.isin(top_genres)],
                x='x', y='y', hue='main_genre', alpha=0.7, s=25)
plt.title('Netflix titles in embedding space (PCA)')
plt.legend(bbox_to_anchor=(1.02, 1), loc='upper left'); plt.show()
```
